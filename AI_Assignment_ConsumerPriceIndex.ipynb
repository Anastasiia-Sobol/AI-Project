```python

# ============================================================
# CPM02 - CONSUMER PRICE INDEX
# Stage 1: Data Exploration
# Project: Predicting Irish CPI
# Platform: Google Colab
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
from io import BytesIO

print("Libraries imported successfully!")


# ============================================================
# 2. UPLOAD AND LOAD DATASET
# ============================================================

print("\nPlease upload your CPI Excel dataset:")

uploaded = files.upload()

# Get uploaded filename
file_name = next(iter(uploaded))

# Read Excel workbook
df = pd.read_excel(
    BytesIO(uploaded[file_name]),
    sheet_name="Unpivoted"
)

print("\nDataset successfully loaded!")


# ============================================================
# 3. EXPLORE DATASET STRUCTURE
# ============================================================

# Display first 10 records
print("\nFirst 10 Records:")
display(df.head(10))

# Dataset dimensions
print("\nDataset Dimensions:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# Column names
print("\nColumn Names:")
print(df.columns.tolist())

# Data types
print("\nData Types:")
print(df.dtypes)

# Dataset information
print("\nDataset Information:")
df.info()

# Statistical summary
print("\nStatistical Summary:")
display(df.describe(include="all"))


# ============================================================
# 4. INVESTIGATE CPI REFERENCE PERIODS
# ============================================================

print("\nAvailable CPI Reference Periods:")

print(
    df["Selected Base Reference Period"].value_counts()
)


# ============================================================
# 5. SELECT DECEMBER 2023 REFERENCE PERIOD
# ============================================================

cpi = df[
    df["Selected Base Reference Period"]
    == "Base December 2023=100"
].copy()

print("\nSelected CPI Dataset:")
display(cpi.head(10))

print("\nSelected Dataset Dimensions:")
print(cpi.shape)


# ============================================================
# 6. INVESTIGATE HISTORICAL COVERAGE
# ============================================================

# Count available CPI observations
print("\nTotal Observations:")
print(cpi["VALUE"].count())

# Find earliest and latest available observations
valid_cpi = cpi.dropna(subset=["VALUE"]).copy()

# Sort before identifying earliest/latest
valid_cpi = valid_cpi.sort_values("TLIST(M1)")

print("\nEarliest Observation:")
print(valid_cpi["Month"].iloc[0])

print("\nLatest Observation:")
print(valid_cpi["Month"].iloc[-1])


# ============================================================
# 7. INVESTIGATE MISSING VALUES
# ============================================================

print("\nMissing Values:")
print(cpi.isnull().sum())

print("\nMissing Value Percentages:")

print(
    (cpi.isnull().sum() / len(cpi)) * 100
)


# ============================================================
# 8. PREPARE DATA FOR VISUALISATION
# ============================================================

# Remove observations without CPI values
cpi = cpi.dropna(subset=["VALUE"]).copy()

# Convert monthly identifiers into datetime
cpi["Date"] = pd.to_datetime(
    cpi["TLIST(M1)"].astype("Int64").astype(str),
    format="%Y%m",
    errors="coerce"
)

# Remove invalid dates, if any
cpi = cpi.dropna(subset=["Date"])

# Sort observations chronologically
cpi = cpi.sort_values("Date").reset_index(drop=True)

# Rename VALUE column for clarity
cpi = cpi.rename(columns={"VALUE": "CPI"})

print("\nPrepared CPI Dataset:")
display(cpi[["Date", "CPI"]].head())


# ============================================================
# 9. CPI STATISTICAL ANALYSIS
# ============================================================

print("\nCPI Statistical Summary:")

display(cpi["CPI"].describe())

print("\nTotal CPI Observations:")
print(len(cpi))


# ============================================================
# 10. VISUALISATION - HISTORICAL CPI
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    cpi["Date"],
    cpi["CPI"],
    label="Consumer Price Index"
)

plt.title("Irish Consumer Price Index (1975-2026)")
plt.xlabel("Year")
plt.ylabel("CPI (December 2023 = 100)")

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 11. CALCULATE ANNUAL INFLATION
# ============================================================

cpi["Inflation"] = (
    cpi["CPI"].pct_change(periods=12) * 100
)

print("\nAnnual Inflation Statistics:")

display(cpi["Inflation"].describe())


# ============================================================
# 12. VISUALISATION - ANNUAL INFLATION
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    cpi["Date"],
    cpi["Inflation"],
    label="Annual Inflation"
)

plt.axhline(
    y=0,
    linestyle="--",
    color="red"
)

plt.title("Annual Inflation Rate in Ireland (1975-2026)")
plt.xlabel("Year")
plt.ylabel("Inflation (%)")

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 13. FINAL DATASET INFORMATION
# ============================================================

print("\nFinal Dataset Information:")

print("Total observations:", len(cpi))
print("Start date:", cpi["Date"].min())
print("End date:", cpi["Date"].max())

print("\nFinal Dataset:")
display(cpi[["Date", "CPI", "Inflation"]].tail(10))


# ============================================================
# END OF STAGE 1 - DATA EXPLORATION
# ============================================================

print("\nStage 1: Data Exploration Completed!")
```
